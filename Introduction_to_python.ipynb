```python
# Arithmetic Operators
a = 10
b = 5

print("Arithmetic Operators:")
print(f"a + b = {a + b}")
print(f"a - b = {a - b}")
print(f"a * b = {a * b}")
print(f"a / b = {a / b}")
print(f"a % b = {a % b}")
print(f"a ** b = {a ** b}")
print(f"a // b = {a // b}")
print()

# Comparison Operators
print("Comparison Operators:")
print(f"a == b: {a == b}")
print(f"a != b: {a != b}")
print(f"a > b: {a > b}")
print(f"a < b: {a < b}")
print(f"a >= b: {a >= b}")
print(f"a <= b: {a <= b}")
print()

# Logical Operators
x = True
y = False

print("Logical Operators:")
print(f"x and y: {x and y}")
print(f"x or y: {x or y}")
print(f"not x: {not x}")
print()

# Assignment Operators
c = 7

print("Assignment Operators:")
c += 3
print(f"c += 3 -> {c}")
c -= 2
print(f"c -= 2 -> {c}")
c *= 2
print(f"c *= 2 -> {c}")
c /= 4
print(f"c /= 4 -> {c}")
c %= 3
print(f"c %= 3 -> {c}")
c **= 2
print(f"c **= 2 -> {c}")
c //= 5
print(f"c //= 5 -> {c}")
print()

# Bitwise Operators
d = 5
e = 3

print("Bitwise Operators:")
print(f"d & e: {d & e}")
print(f"d | e: {d | e}")
print(f"d ^ e: {d ^ e}")
print(f"~d: {~d}")
print(f"d << 1: {d << 1}")
print(f"d >> 1: {d >> 1}")
```

```python
def count_steps(n):
    if n==0:
        return 0
    if n==1:
        return 1
    if n==2:
        return 2
    a=1
    b=2
    for _ in range(3,n+1):
        a,b=b,a+b
        print(a,b)
    return b

staircase=int(input("Enter number of staircase"))
a=count_steps(staircase)
print(f"number of ways: {a}")
```

```python
def pattern1(num_stars):

    if num_stars % 2 != 0:
        print("Please enter an even number.")
    else:

        for i in range(num_stars // 2):
            print(" " * i + "*" * (num_stars - (2 * i)))

        print(" " * (num_stars // 2) + "*")

def pattern2(num_rows):
    for i in range(1, num_rows + 1):
        if i % 2 != 0:
            print(" ** ")
        else:
            print("******")

num_stars = int(input("Enter the number of stars for the first pattern: "))
pattern1(num_stars)

num_rows = int(input("Enter the number of rows for the second pattern: "))
pattern2(num_rows)

```
